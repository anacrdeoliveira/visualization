My personal notes preparing for class and assignment 2

We learned about the three important qualities of data visualization:
> Aesthetic: it is pleasing to look at

> Substantive: it presents data accurately and honestly

> Perceptual: the message its maker is trying to convey is easy to understand



We also need to think about what the visualization needs to accomplish: the intended purpose, the intended audience, and the intended medium/media. 

Kennedy et al. (2016) defended 4 conventions for data visualization: two-dimensional images, clean layouts, geometric shapes and lines, and inclusion of data sources (at the bottom of the image). 

Citing sources of data is good for reproductibility, but also to help audience trust data.
