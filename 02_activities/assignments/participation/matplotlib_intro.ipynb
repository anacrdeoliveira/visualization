```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

Making sample data, so I can play with plotting

```python
# making random data
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
```

```python
# scatter plotting
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
# bar plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
```

```python
# line plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
fig.tight_layout() ## added in class
```

```python
# turning it into an histogram
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y, color="purple") # I added color="purple" for fun
fig.tight_layout()
```

Adding titles to the graph:

```python
# step 1 - setting labels
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
```

```python
# step 2 - choosing font and style so I can use fontdict later
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}
```

```python
# add fontdict arguments to our previous lineplot
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

```python
# moving labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
```

Remaking the scatterplot, but this time adding color and marker arguments 

```python
# very cool transformation into purple stars 
# (indigo shows as purple for me)
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter( x,
          y,
          marker='*',         # NEW ADDITIONS
          color = "indigo")   # NEW ADDITIONS
fig.show()
fig.tight_layout()
```

```python
# adjusting style and width
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='*', 
        color = 'indigo', 
        linestyle = '--',   # NEW ADDITIONS
        linewidth = 2)      # NEW ADDITIONS
fig.show()
fig.tight_layout()
```

```python
# modifying color using hex codes
fig, ax = plt.subplots(figsize=(5, 3))
# color being the new addition
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
fig.show()
fig.tight_layout()
```

Customizing the plot further:

```python
ax.plot(x,y,marker='*',
        markersize = 12,              # NEW ADDITIONS
        color = '#7425b9',
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',  # NEW ADDITIONS
        markerfacecolor = '#fa9359' ) # NEW ADDITIONS
```

```python
# plotting to see it
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y,marker='*',
        markersize = 12,              # NEW ADDITIONS
        color = "red",
        linestyle = '--',
        linewidth = 2,
        markeredgecolor = '#fa9359',  # NEW ADDITIONS
        markerfacecolor = "yellow" ) # NEW ADDITIONS
fig.show()
```

Adding grid lines, then modifying the design of the grid lines

Trying to plot again (not in slides)

```python
# modifying my grid and adding it to the graph
fig, ax = plt.subplots(figsize=(5, 3))
# color being the new addition
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
ax.grid(axis = 'y', color='#F4C2C2',linestyle='-',linewidth=1)
ax.grid(axis = 'x', color='#89CFF0',linestyle='-',linewidth=1)
fig.show()
fig.tight_layout()

# #89CFF0 = baby blue
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
# changing colors again
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2, markeredgecolor = '#fa9359', markerfacecolor = "yellow")
ax.grid(axis = 'y', color='#F4C2C2',linestyle='-',linewidth=1)
ax.grid(axis = 'x', color='#89CFF0',linestyle='-',linewidth=1)
fig.show()
fig.tight_layout()
```

Final activity - I chose "Contour Plot using Seaborn"

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('iris')
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length, cmap="Reds", fill=True)
plt.show()
```

```python
sns.set_style("white")
sns.kdeplot(x=df.sepal_width, y=df.sepal_length,
            cmap="Blues", fill=True, bw_adjust=0.5)
plt.show()
```
