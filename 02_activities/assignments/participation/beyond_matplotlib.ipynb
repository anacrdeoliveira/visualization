```python
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
# Loading tips sample dataset
tips = sns.load_dataset("tips")
print(tips)
```

```python
# Tip amount versus total bill
sns.lineplot(data=tips,       # choose our dataset
           x='total_bill',  # define our x variable
           y='tip')         # define our y variable
```

```python
# Trying different styles (I liked darkgrid)
# 'darkgrid', 'whitegrid', 'dark', 'white', 'ticks'
sns.set_style('darkgrid')
sns.lineplot(data=tips,       # choose our dataset
           x='total_bill',  # define our x variable
           y='tip')     # define our y variable
```

```python
# Adding axis title and labels

tipgraph = sns.lineplot(data=tips,
                        x='total_bill',
                        y='tip')
tipgraph.set(title='Tips vs. Total Bill', 
            xlabel='Total Bill ($)', 
            ylabel='Tip Amount ($)')
```

```python
# Changing size of the plot

fig = plt.subplots(figsize=(10, 3))
```

```python
tipgraph = sns.lineplot(data=tips,
                      x='total_bill',
                      y='tip',
                      color = 'hotpink',
                      linestyle = '--',
                      linewidth = 3,
                      marker = 'o',
                      markerfacecolor = 'indigo')
```

```python
tipgraph = sns.scatterplot(data=tips, x='total_bill',
                        y='tip', style = 'time', hue = 
                        'day', palette = ['purple',
                        'hotpink', 'deepskyblue',
                        'yellowgreen'])
tipgraph.set(title='Tips vs. Total Bill',
            xlabel='Total Bill ($)',
            ylabel='Tip Amount ($)')
```

```python
# so cool!!
sns.pairplot(
      data = tips,
      hue = 'day')
```

```python
daysplot = sns.relplot(
            data=tips,
            x="total_bill",
            y="tip",
            hue="sex",
            col="day",
            kind="scatter",
            col_wrap=2)
```

```python
# All in one chunk

import plotly.graph_objects as go     # 'go' is 'graph objects’
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])

graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))
graph.update_layout(
    title="Pirate Scores",
    xaxis_title="Pirates",
    yaxis_title="Score")

graph.show()
```

```python
# Wordclouds
import wordcloud
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv",
on_bad_lines='skip')
df
```

```python
# example from slides:

# join all our text from each row from our quote column into a string
text = " ".join(each for each in df.quote)
# generate our wordcloud image
wordcloud = WordCloud(background_color="black", # I changed from white to black
colormap = 'inferno').generate(text)
# use matplotlib syntax to put our image in a figure
fig, ax = plt.subplots(figsize=(7, 3))
ax.imshow(wordcloud,                  # remember 'imshow' from when we added pictures to our matplotlib axes
        interpolation='bilinear')   # this line helps smooth our image
ax.axis("off")
```

```python
# Venn diagrams
from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
```

```python
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
diagram = venn2_unweighted([A, B],
                            set_labels = ('Fruits', 'Pies'),
                            set_colors=("blue", "red"),
                            alpha=0.5)
plt.show()
```

```python
# Adding names instead of numbers:

A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
diagram = venn2_unweighted([A, B],
                            set_labels = ('Fruits', 'Pies'),
                            set_colors=("blue", "red"),
                            alpha=0.5)
diagram.get_label_by_id("10").set_text(" ".join(A - B)) # 10 means A
diagram.get_label_by_id("11").set_text(" ".join(A & B)) # 11 means A and B
diagram.get_label_by_id("01").set_text(" ".join(B - A)) # 01 means B
# If the venn had A, B, and C, we could use 100, 010, and 001, etc!

plt.show()

```
