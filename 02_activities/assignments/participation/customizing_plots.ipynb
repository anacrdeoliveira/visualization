```python
# Libraries (05_customizing_our_plots.pdf)

import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

```python
# My first attempt to plot both variables on the same graph (axes)
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1,y2)
fig.show()
fig.tight_layout()
```

```python
# The code from the slides 
# why not put y1 and y2 on the same line?
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
```

```python
# Add a legend
# Now I can see why y1 and y2 on the same line won't work
# at least not when using "label"

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='lower right')
fig.show()
```

```python
# Changing elements on the legend:
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='lower right',
        frameon = True, #add frame around the legend
        fontsize = 12, #change font size
        ncol = 2, #specify number of columns
        shadow = True )
fig.show()
fig.tight_layout # I added this one
```

```python
# Putting the legend outside the plot area
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',
        bbox_to_anchor =(1, 1)) # I played a little, (0,1) was cool, but over the plot
fig.show()
fig.tight_layout
```

```python
# Putting data on a scatter plot:
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
# testing text function:
ax.text(10, 95, "This value is important!")
#fig.show() # necessary? it plots without it
fig.tight_layout # added
```

```python
# modifying text annotations
# add to the plot code otherwise nothing shows
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.text(10, 95, "This value is important!",
      ha='center',    ##alignment of text
      color = 'red',  ##modify font colour
      size = 20)      ##modify font size
fig.show()
```

```python
# Positioning text annotations - step 1
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
```

```python
# Part 2 - adding text to positions within that matrix
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5, ". Data:(1, 5)",           transform=ax.transData    )
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)",   transform=ax.transAxes    )
ax.text(0.2, 0.2, ". Figure:(0.2, 0.2)", transform=fig.transFigure )
```

```python
# Point and annotate
# requires location of the point and location of the text
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!', xy=(10, 95), xytext=(20, 94),
          arrowprops=dict(facecolor='purple')) # I changed the color
fig.show()
```

```python
# Arrow styles "wedge"
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "wedge",
                            color = "hotpink"))
fig.show()
```

```python
# Arrow styles ->
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "->",
                            color = "hotpink"))
fig.show()
```

```python
# Arrow styles <->
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "<->",
                            color = "hotpink"))
fig.show()
```

```python
# Arrow styles -[
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "-[",
                            color = "hotpink"))
fig.show()
```

```python
# Arrow styles ]-[]
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "]-[",
                            color = "hotpink"))
fig.show()
```

```python
# Arrow styles "fancy"
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "fancy",
                            color = "hotpink"))
fig.show()
```

```python
#removing ticks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.yaxis.set_major_locator(plt.NullLocator()) 
ax.xaxis.set_major_formatter(plt.NullFormatter()) # removes only labels
```

```python
#limiting maximum tick marks on the x axis
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(3))
```

```python
#customizing intervals
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
#rotating axis
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right')
```

```python
#changing font size (reducing to 5)
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right',size=5)
```

```python
#using a title with a specific font
font1 = {'family':'serif','color':'indigo'} #corrected a space here
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xlabel("Shiny New X Axis!", fontsize = 18, fontdict = font1)
```
