I downloaded a csv file containing data, converted it into excel, and went over the following steps to produce an histogram. The histogram is inside the excel file, on the tab "Histogram", and also saved as a "Assigment-3-histogram.png".

>1. I selected all data and created a pivot table

>2. I selected "Year", "Geography", "Type of Work", "Age Group", and "Wages" as filters. The filters were set as "2016", "Ontario", "Full Time", "15 years or over" and "Average hourly wage rate".

>3. I useed "Educational Level" as rows, "Sex" as columns, and the average of "Value" to be displayed.

>4. I copied the resulting table (see tab "Filtering data") and pasted it on the "Histogram" tab, so I could produce the graphs.

>5. Before plotting the histogram I excluded the "Grand Total" row because the dataset already comes with the total average of wages, so the result is inacurate

>6. I also eliminated the row with "Total, all education levels", because it represents an overall average I don't need to emphasize, and deleted unecessary spaces on names. 

>7. Finally, I ordered the values from the lowest to the highest.

> What software did you use to create your data visualization?

I used Excel (pivot table, histogram)

> Who is your intended audience? 

Labour Economics students, or any people trying to understand how wage varies based on education and gender.
    
> What information or message are you trying to convey with your visualization? 

The human capital model predicts that higher wages should be expected for higher levels of education, and that is true. However, the human capital model does not explain why women with the same level of education still receive lower wages than men. In this graph, I wanted to show that even if all workers are full time employed and have the same level of education, the blue bar representing men is always bigger than the pink bar representing women.

> What aspects of design did you consider when making your visualization? How did you apply them? With what elements of your plots? 

I considered the kind of information a histogram could show, the side by side bar for each gender grouped according to the level of education on the x-axis. I also ordered it in a way that makes sense for the data (from the lowest to the highest), which is the expected trend, so the observer can also notice the trend and verify that educational levels and wages grow "together". I chose filters that help understanding (e.g., hourly wages are well know in Ontario, by the labour economics student as well as by any person), and that make data less confusing (2016 only, the most recent data available). Finally, I chose a title that summarizes the graph information, and put the male and female labels up and centre.

I explained the steps I took. The title summarizes the data sources and "filters" I used. Any person with access to the data could find it. I could improve it by providing the link to the data, but that information would make the graph less visually appealing.

> How did you ensure that your data visualization is accessible? 

I used high contrast colors that also have different strengths, and used the same order for all the items (female on the left, automatically generated), so it can be understood by color blind people. Some of the labels on the x axis are not readable, but they follow a trend that is easy to understand and keep accuracy with the original data, so I chose to leave them as they were.

> How did you choose which features of your chosen dataset to include or exclude from your visualization? 

I used logic and experience, but also personal related elements. For the year, I went with the most recent available data. I chose to focus on Ontario, the province where we are at, and where the data was provided. I chose hourly wage instead of weekly because it is more common in my field and in public understanding. I used full time workers because there are historical assumptions on whether the gender wage gap might be explained by the fact that women more often would have part time positions compared to men, and, for the same reason, the choice of an averae wage makes sense. I decided to keep the broad range of all ages above 15 to be more generic (older employees are expected to make more money on average)
